## Investigation for [Booking ID]

```python
total_rows = len(df)
unique_ids = df['Booking ID'].nunique()
duplicate_ids = df['Booking ID'].duplicated().sum()

print("Total rows:", total_rows)
print("Unique Booking IDs:", unique_ids)
print("Duplicate rows (ID level):", duplicate_ids)
```

```python
dup_id_counts = (
    df['Booking ID'].value_counts().loc[lambda x: x > 1]
)

print("Number of duplicated IDs:", len(dup_id_counts))
dup_id_counts.head()
```

```python
problem_id = dup_id_counts.index[0]
df[df['Booking ID'] == problem_id]
```

#### Result
Apparently, there is a problem with the booking system as there is no relation between the the records with duplicate Booking ID. Later, in the data refining we should think about a solution.
-  One solution could be to check whether the combination of BookingID and CustomerID are unique or no. If yes, this can be used as a Primary Key.


## Logical Relation Between Fields

#### Booking ID

* Should represent the **primary key at booking-level grain**
* If duplicated → Need to be fiexed later.

#### Customer ID

* One-to-many relationship with Booking ID
* A customer can create multiple bookings (but not many did lol)
* Should never change within the same Booking ID

#### Date + Time

* Together define the **request timestamp**
* If Booking ID appears multiple times with different Time → dataset is tracking state transitions
* If Time changes but Booking ID same → this is likely **status evolution**, not duplicate booking

#### Booking Status

* All cancellation and rating fields depend on this column
* 
#### Cancelled Rides by Customer | Reason for cancelling by Customer

Logical rule:

* If Cancelled Rides by Customer > 0
  → Booking Status must be "Cancelled"
  → Reason must NOT be null

If:

* Status ≠ Cancelled
* but cancellation fields are populated
  → Data inconsistency

#### Cancelled Rides by Driver | Driver Cancellation Reason

Same dependency rule:

* If Cancelled Rides by Driver > 0
  → Status = Cancelled
  → Driver Cancellation Reason must exist

Important:
Only one side should cancel per booking.

If both customer and driver cancellation > 0 → data integrity issue.

#### Incomplete Rides | Incomplete Rides Reason

Logical relation:

* If Incomplete Rides > 0
  → Status = Incomplete
  → No Customer Rating should exist
  → Booking Value likely = 0 or partial

#### Completion Logic

If Booking Status = Completed:

Then:

* Ride Distance > 0
* Booking Value > 0
* Driver Ratings not null (but could be)
* Customer Rating not null (but could be)
* No cancellation fields populated

If any of these missing → incomplete completion data.

#### Booking Value

Logically depends on:

* Ride Distance
* Vehicle Type
* Possibly surge pricing (not shown here)

If:

* Ride Distance = 0
* Booking Value > 0
  → anomaly

#### Vehicle Type

Logical behavior:

* Impacts Booking Value
* Impacts Ride Distance pricing
* Does NOT change within same Booking ID

If duplicated Booking ID has different Vehicle Type → structural issue (which has)

#### Customer Rating | Driver Ratings

Logical rule:

* Only exist if ride completed
* Should probably be null if:

  * Cancelled
  * Incomplete

#### Payment Method

Logical behavior:

* Should only exist if ride completed
* Should be null for cancelled rides

If payment method exists but Booking Value = 0 → anomaly
